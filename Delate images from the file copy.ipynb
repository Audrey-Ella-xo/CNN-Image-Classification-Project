```python
import os

def rename_images_in_subfolders(root_folder):
    if not os.path.isdir(root_folder):
        print("The specified path is not a valid folder.")
        return
    
    image_extensions = ("jpeg", "jpg", "png", "bmp", "gif")
    
    for subdir, _, files in os.walk(root_folder):
        folder_name = os.path.basename(subdir)
        images = [f for f in files if f.lower().endswith(image_extensions)]
        
        # Find the highest existing index to continue numbering
        existing_indices = []
        for image in images:
            parts = os.path.splitext(image)  # Split name and extension
            name_parts = parts[0].split('_')
            if len(name_parts) > 1 and name_parts[0] == folder_name:
                try:
                    existing_indices.append(int(name_parts[1]))
                except ValueError:
                    continue  # Ignore files that don't match the expected format
        
        start_index = max(existing_indices) if existing_indices else 0
        
        # Rename only new files
        for image in images:
            old_path = os.path.join(subdir, image)
            ext = os.path.splitext(image)[1]  # Get file extension
            
            # Skip renaming if the file is already properly named
            if image.startswith(folder_name + "_"):
                continue
            
            # Generate a new name that doesn't overwrite existing files
            start_index += 1
            new_name = f"{folder_name}_{start_index}{ext}"
            new_path = os.path.join(subdir, new_name)
            
            os.rename(old_path, new_path)
            print(f"Renamed in {folder_name}: {image} -> {new_name}")

# Change this to your actual dataset folder
root_folder = '/Users/Audrey_xo/Library/CloudStorage/GoogleDrive-anitaudrey@gmail.com/.shortcut-targets-by-id/1uZ3cbI9IgIvVi6ppt9wv_Ntlu9zlcVha/Colab Notebooks/Images DLGA'
rename_images_in_subfolders(root_folder)
```
