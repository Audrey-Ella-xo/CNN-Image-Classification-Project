```python
import os
import random

# Define dataset path
# dataset_path = "..."  # Modify this to your dataset path
dataset_path = '/Users/Audrey_xo/Library/CloudStorage/GoogleDrive-anitaudrey@gmail.com/.shortcut-targets-by-id/1uZ3cbI9IgIvVi6ppt9wv_Ntlu9zlcVha/Colab Notebooks/Images DLGA'

# Allowed image formats
image_extensions = {"jpeg", "jpg", "png", "bmp", "gif"}

# Target image count per class
target_image_count = 1000

missing_folders = []
excess_folders = []

for class_folder in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_folder)
    
    if not os.path.isdir(class_path):
        continue
    
    images = os.listdir(class_path)
    
    # # Remove invalid format images
    # images_to_remove = [img for img in images if img.split('.')[-1].lower() not in image_extensions]
    # for img in images_to_remove:
    #     os.remove(os.path.join(class_path, img))
    # print(f"Removed {len(images_to_remove)} invalid images in {class_folder}.")
    
    # Count valid images
    valid_images = [img for img in images if img.split('.')[-1].lower() in image_extensions]
    num_images = len(valid_images)
    
    # Compare with target image count
    if num_images < target_image_count:
        missing_folders.append((class_folder, target_image_count - num_images))
    elif num_images > target_image_count:
        excess_folders.append((class_folder, num_images - target_image_count))

# Report results
if missing_folders:
    print('\nFolders with fewer than the target number of images:')
    for folder, missing in missing_folders:
        print(f"- {folder}: {missing} images missing.")
else:
    print("\nNo folders are missing images.")

if excess_folders:
    print('\nFolders with more than the target number of images:')
    for folder, excess in excess_folders:
        print(f"- {folder}: {excess} extra images.")
else:
    print("\nNo folders have excess images.")

# Final check
print("\nImage validation and comparison completed.")
```
